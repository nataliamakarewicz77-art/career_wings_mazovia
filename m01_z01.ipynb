```python
#TERAZ 3 RÓŻNE SPOSOBY NA ZAPISANIE TEGO SAMEGO
imie = "Jagoda"
nazwisko = "Kowalska"
wiek = 22

# SPOSÓB 1: f-string (formatted string literal) - Python 3.6+
# Zmienne wstawiane są bezpośrednio w nawiasach klamrowych {} wewnątrz stringa
# Najbardziej czytelny i nowoczesny sposób
#NAJLEPSZY PRZY DUŻEJ ILOŚCI DANYCH
print(f"{imie} {nazwisko} ma {wiek} lat")

# SPOSÓB 2: print z wieloma argumentami oddzielonymi przecinkami
# Python sam dodaje spacje między argumentami i konwertuje wiek (int) na tekst
# Nie trzeba ręcznie łączyć stringów ani konwertować typów
print(imie, nazwisko, "ma", wiek, "lat")

# SPOSÓB 3: formatowanie operatorem % (stary styl, znany z języka C)
# %s = miejsce na string, %d = miejsce na liczbę całkowitą (int)
# Wartości podaje się w krotce (tuple) po znaku %
print("%s %s ma %d lat" % (imie, nazwisko, wiek))
```

```python
#ZADANIE DOMOWE: Napisz skrypt, który pyta użytkownika o imię i rok urodzenia, oblicza ile lat zostało do emerytury i wypisuje spersonalizowaną wiadomość.

imie = input("Podaj imię: ")
rok_urodzenia = int(input("Podaj rok urodzenia: "))
#int konwertuje wartość ze string na integer
#przyjmujemy, że:
wiek_emerytalny = 65
aktualny_rok = 2026
lata_do_emerytury = (wiek_emerytalny - aktualny_rok + rok_urodzenia)
if lata_do_emerytury < 0:
    print(f"{imie}, już jesteś na emeryturze już {lata_do_emerytury * -1} lat")
elif lata_do_emerytury == 0:
    print(f"{imie}, właśnie przechodzisz na emeryturę")
else:
    print(f"Cześć {imie}, zostało Ci {lata_do_emerytury} lat do emerytury.")
```
