```python
#ZADANIE DOMOWE: Napisz skrypt, który pyta użytkownika o imię i rok urodzenia, oblicza ile lat zostało do emerytury i wypisuje spersonalizowaną wiadomość.

imie = input("Podaj imię: ")
rok_urodzenia = int(input("Podaj rok urodzenia: "))
#int konwertuje wartość ze string na integer
#przyjmujemy, że:
wiek_emerytalny = 65
aktualny_rok = 2026
lata_do_emerytury = (wiek_emerytalny - aktualny_rok + rok_urodzenia)
if lata_do_emerytury < 0:
    print(f"{imie}, już jesteś na emeryturze już {lata_do_emerytury * -1} lat")
elif lata_do_emerytury == 0:
    print(f"{imie}, właśnie przechodzisz na emeryturę")
else:
    print(f"Cześć {imie}, zostało Ci {lata_do_emerytury} lat do emerytury.")
```
